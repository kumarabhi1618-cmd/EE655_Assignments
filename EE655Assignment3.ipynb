```python
# Required libraries
import numpy as np
```

```python
# Spatial Pyramid Pooling for One Feature Map
def spatial_pyramid_pooling(feature_map, pyramid_levels=[1, 2, 4, 8]):
    """
    Applies Spatial Pyramid Pooling (SPP) using max pooling
    on a single feature map.

    Parameters:
    feature_map : 2D NumPy array (H × W)
    pyramid_levels : list of pyramid divisions

    Returns:
    1D NumPy array containing pooled features
    """

    H, W = feature_map.shape
    spp_features = []

    # Loop over pyramid levels
    for level in pyramid_levels:

        # Size of each spatial bin
        bin_h = H / level
        bin_w = W / level

        # Divide feature map into (level × level) bins
        for i in range(level):
            for j in range(level):

                # Compute bin boundaries
                h_start = int(i * bin_h)
                h_end   = int((i + 1) * bin_h)
                w_start = int(j * bin_w)
                w_end   = int((j + 1) * bin_w)

                # Extract region
                region = feature_map[h_start:h_end, w_start:w_end]

                # Apply max pooling
                spp_features.append(np.max(region))

    return np.array(spp_features)
```

**What this does**

Divides the feature map into spatial bins

Applies max pooling in each bin

Collects pooled values into a 1D vector

Output size depends only on pyramid levels

```python
# Spatial Pyramid Pooling for RGB Image
def spp_rgb_image(image):
    """
    Applies SPP on an RGB image by treating
    each color channel as a feature map.

    Parameters:
    image : H × W × 3 RGB image

    Returns:
    1D fixed-length SPP feature vector
    """

    spp_vector = []

    # Process each color channel independently
    for channel in range(3):  # R, G, B
        channel_map = image[:, :, channel]
        channel_features = spatial_pyramid_pooling(channel_map)
        spp_vector.extend(channel_features)

    return np.array(spp_vector)
```

**Explanation**

RGB image → 3 separate feature maps

Apply SPP on each channel

Concatenate all pooled values

Output length = 255

```python
# Demonstration with Different Image Sizes

# Create two RGB images of different sizes
image_1 = np.random.randint(0, 256, (128, 128, 3), dtype=np.uint8)
image_2 = np.random.randint(0, 256, (300, 180, 3), dtype=np.uint8)

# Apply Spatial Pyramid Pooling
features_1 = spp_rgb_image(image_1)
features_2 = spp_rgb_image(image_2)

# Print results
print("Image 1 shape:", image_1.shape)
print("Image 2 shape:", image_2.shape)
print("Feature vector 1 length:", features_1.shape)
print("Feature vector 2 length:", features_2.shape)
```

**Why SPP Works (Short Theory)**

CNNs normally require fixed-size inputs

SPP removes this limitation by:

Dividing feature maps into adaptive spatial bins

Pooling each bin independently

Output length depends on number of bins, not image size


**Key point**

Spatial Pyramid Pooling converts variable-sized feature maps into a fixed-length representation by applying pooling over multi-level spatial partitions
